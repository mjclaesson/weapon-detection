```python
def get_metrics(filename, model_name):
    df = pd.read_csv(filename)
    # idx = df["metrics/mAP50(B)"].idxmax()
    # idx = df["metrics/precision(B)"].idxmax()
    idx = df["metrics/mAP50-95(B)"].idxmax()
    
    precision = df.loc[idx, "metrics/precision(B)"]
    recall = df.loc[idx, "metrics/recall(B)"]
    map50 = df.loc[idx, "metrics/mAP50(B)"]
    map50_95 = df.loc[idx, "metrics/mAP50-95(B)"]

    print(model_name)
    print(f"Idx: {idx}")
    print(f"Precision: {precision}")
    print(f"Recall: {recall}")
    print(f"mAP50: {map50}")
    print(f"mAP50-95: {map50_95}")
    print("")

    return precision, recall, map50, map50_95

a = get_metrics("results_yolo5nu.csv", "YOLO 5n")
b = get_metrics("results_8n.csv", "YOLO 8n")
c = get_metrics("results_11n.csv", "YOLO 11n")
d = get_metrics("results_26n.csv", "YOLO 26n")
```

# RESNET50

```python
def get_resnet_metrics(filename, model_name):
    df = pd.read_csv(filename)
    idx = df["val_loss"].idxmin()
    
    precision = df.loc[idx, "val_precision"]
    recall = df.loc[idx, "val_recall"]
    accuracy = df.loc[idx, "val_acc"]

    print(model_name)
    print(f"Index: {idx}")
    print(f"Accuracy: {accuracy}")
    print(f"Precision: {precision}")
    print(f"Recall: {recall}")
    print("")

    return precision, recall

a = get_resnet_metrics("results_1.csv", "RESNET50 Fine tune last conv")
b = get_resnet_metrics("results_64_3layers.csv", "RESNET50 Fine tune last 3 conv")
```

# Confusion Matrix: Classification Conversion

- weapon: gun or knife
- no weapon: background

- TP: predicted weapon AND true weapon
- FP: predicted weapon BUT true background
- FN: predicted background BUT true weapon
- TN: predicted background AND true background

- Precision = TP / (TP + FP)
- Recall = TP / (TP + FN)


```python
def get_precision(TP, FP): 
    return TP / (TP + FP)
def get_recall(TP, FN): 
    return TP / (TP + FN)
def get_accuracy(TP, TN, FP, FN): 
    return (TP + TN) / (TP + TN + FP + FN)
```

```python
validation_size = 1800

# right up: yolo 5
print("yolo v5nu")
TP = (689+1+3+294)
FP = (216+80)
FN = (194+27)
TN = validation_size - (TP + FP + FN)
P = get_precision(TP, FP)
R = get_recall(TP, FN)
A = get_accuracy(TP, TN, FP, FN)
print(f"Accuracy: {A}")
print(f"Precision: {P}")
print(f"Recall: {R}")
print()

# right down: yolo 8
print("yolo 8n")
TP = (691+2+6+297)
FP = (208+74)
FN = (189+23)
TN = validation_size - (TP + FP + FN)
P = get_precision(TP, FP)
R = get_recall(TP, FN)
A = get_accuracy(TP, TN, FP, FN)
print(f"Accuracy: {A}")
print(f"Precision: {P}")
print(f"Recall: {R}")
print()

# left down: yolo 11
print("yolo 11n")
TP = (699+1+2+296)
FP = (218+69)
FN = (185+25)
TN = validation_size - (TP + FP + FN)
P = get_precision(TP, FP)
R = get_recall(TP, FN)
A = get_accuracy(TP, TN, FP, FN)
print(f"Accuracy: {A}")
print(f"Precision: {P}")
print(f"Recall: {R}")
print()

# left up: yolo 26
print("yolo 26n")
TP = (658+1+6+297)
FN = (159+49)
FP = (222+24)
TN = validation_size - (TP + FP + FN)
P = get_precision(TP, FP)
R = get_recall(TP, FN)
A = get_accuracy(TP, TN, FP, FN)
print(f"Accuracy: {A}")
print(f"Precision: {P}")
print(f"Recall: {R}")
print()
```
